```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        
    def constructTree(self, arr ,root, i, n):
        if i<n:
            root = TreeNode(arr[i-1])
            root.left = self.constructTree(arr, root.left, 2*i, n)
            root.right = self.constructTree(arr, root.right, 2*i+1, n)
        return root
        
    def preOrder(self, root, result):
        if not root: return []
        result.append(root.value)
        self.preOrder(root.left, result)
        self.preOrder(root.right, result)
        return result
        
    def inOrder(self, root, result):
        if not root: return []
        self.inOrder(root.left, result)
        result.append(root.value)
        self.inOrder(root.right, result)
        return result
        
    def postOrder(self, root, result):
        if not root : return []
        self.postOrder(root.left, result)
        self.postOrder(root.right, result)
        result.append(root.value)
        return result
```

### Binary Search Trees

```python
class TreeNode:
    def __init__(self, val=None):
        self.val=val
        self.left=None
        self.right=None

    def constructTree(self,root,key):
        if not root:
            return TreeNode(key)
        if key>root.val:
            root.right=self.constructTree(root.right,key)
        else:
            root.left=self.constructTree(root.left,key)
        return root

    def preOrder(self,root,res=[]):
        if not root:
            return [3]
        res.append(root.val)
        self.preOrder(root.left,res)
        self.preOrder(root.right,res)
        return res
        
    def inOrder(self,root,res=[]):
        if not root: return []
        self.inOrder(root.left,res)
        res.append(root.val)
        self.inOrder(root.right,res)
        return res
    
    def postOrder(self,root,res=[]):
        if not root:
            return []
        self.postOrder(root.left,res)
        self.postOrder(root.right,res)
        res.append(root.val)
        return res

    def levelOrder(self,root,res=[]):
        if not root: return []
        queue = [root]
        while queue:
            node=queue.pop(0)
            res.append(node.val)
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        return res

    def height(self,root):
        if not root: return 0
        return 1 + max(self.height(root.left),self.height(root.right))

    def search(self,root,ele):
        if not root: 
            print("no root :: ")
            return False
        if root.val==ele:
            print("matches ")
            return True
        return self.search(root.left, ele) or self.search(root.right,ele)

arr = [8, 3, 10, 1, 6, 14, 4, 7]
root = None
obj=TreeNode()
for val in arr:
    root = obj.constructTree(root, val)
print(root)
```
