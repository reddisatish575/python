```python
class ListNode:
    def __init__(self, val):
        self.val = val
        self.next = None

class LinkedList:
    def __init__(self):
        self.head=None

    def generateLL(self,arr):
        if not self.head:
            self.head=ListNode(arr[0])
        itr=self.head
        for i in range(1,len(arr)):
            node=ListNode(arr[i])
            itr.next=node
            itr=itr.next
        return self.display()

    def display(self):
        lstr=''
        itr=self.head
        while itr:
            lstr+=str(itr.val)+'-->' if itr.next else str(itr.val)
            itr=itr.next
        return lstr

    def sorting(self):
        itr = self.head
        while itr:
            min_value=itr
            temp = itr.next
            while temp:
                if temp.val<min_value.val:
                    min_value=temp
                temp=temp.next
            itr.val,min_value.val=min_value.val,itr.val
            itr=itr.next
        return self.display()
obj=LinkedList()
arr = [3,6,2,6,3,8,32,9]
obj.generateLL(arr)
obj.sorting()
```
