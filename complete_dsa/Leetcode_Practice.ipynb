```python
class Solution:
    def lengthOfLongestSubstring(self, s):
        unique_elements = set()
        left = 0
        res = 0
        for i in range(len(s)):
            if s[i] not in unique_elements:
                unique_elements.add(s[i])
                res = max(res, i-left+1)
            else:
                while s[i] in unique_elements:
                    unique_elements.remove(s[left])
                    left+=1
                unique_elements.add(s[i])
                res = max(res, i-left+1)
        return res
obj=Solution()
obj.lengthOfLongestSubstring("abcabcbb")
```
