```python
class BubbleSort:
    def bubbleSort(self,arr):
        if not arr : return []
        l=len(arr)
        for i in range(l-1):
            isSwapped=False
            for j in range(l-i-1):
                if arr[j]>arr[j+1]:
                    arr[j],arr[j+1]=arr[j+1],arr[j]
                    isSwapped=True
            if not isSwapped:
                break
        return arr
if __name__=='__main__':
    obj=BubbleSort()
    testCases = [[], [0,0],[-1,1,1],[201,492,6,1,0,1,0,3,6],[-1,-7,-1,0,-3,-6],[9,4,-1,5,2,-7]]
    for c,i in enumerate(testCases):
        print("Test Case -",c," ",i, "-->", obj.bubbleSort(i)) 
```

```python
class InsertionSort:
    def insertionSort(self,arr):
        l=len(arr)
        for i in range(1,l):
            value=arr[i]
            gap=i
            while gap>0 and arr[gap-1]>value:
                arr[gap]=arr[gap-1]
                gap-=1
            if(gap!=i):
                arr[gap]=value
        return arr
if __name__=='__main__':
    obj=InsertionSort()
    testCases = [[], [0,0],[1,1,3,2,-1,-1],[201,492,6,1,0,1,0,3,6],[-1,-7,-1,0,-3,-6],[9,4,-1,5,2,-7]]
    for c,i in enumerate(testCases):
        print("Test Case -",c," ",i, "-->", obj.insertionSort(i)) 
```

```python
class SelectionSort:
    def selectionSort(self,arr):
        l=len(arr)
        for i in range(l-1):
            minIndex=i
            for j in range(i+1, l):
                if(arr[j]<arr[minIndex]):
                    minIndex=j
            if(minIndex!=i):
                arr[i],arr[minIndex]=arr[minIndex],arr[i]
        return arr
if __name__=='__main__':
    obj=SelectionSort()
    testCases = [[], [0,0],[1,1,3,2,-1,-1],[201,492,6,1,0,1,0,3,6],[-1,-7,-1,0,-3,-6],[9,4,-1,5,2,-7]]
    for c,i in enumerate(testCases):
        print("Test Case -",c," ",i, "-->", obj.selectionSort(i)) 
```

```python
class MergeSort:
    def divide(self,arr):
        l = len(arr)
        if l<=1:
            return arr
        mid=l//2
        left=self.divide(arr[:mid])
        right=self.divide(arr[mid:])
        return self.conquer(left,right)

    def conquer(self,left,right):
        i,j=0,0
        res=[]
        while i<len(left) and j<len(right):
            if left[i]<right[j]:
                res.append(left[i])
                i+=1
            else:
                res.append(right[j])
                j+=1
        res.extend(left[i:])
        res.extend(right[j:])
        return res    
if __name__=='__main__':
    obj=MergeSort()
    testCases = [[], [0,0],[1,1,3,2,-1,-1],[201,492,6,1,0,1,0,3,6],[-1,-7,-1,0,-3,-6],[9,4,-1,5,2,-7]]
    # testCases = [[9,4,-1,5,2,-7]]
    for c,i in enumerate(testCases):
        print("Test Case -",c," ",i, "-->", obj.divide(i)) 
```

```python
class QuickSort:
    def quickSort(self,arr,left,right):
        if left<right:
            pos=self.partition(arr,left,right)
            self.quickSort(arr,left,pos-1)
            self.quickSort(arr,pos+1,right)
        return arr

    def partition(self,arr,left,right):
        i=left
        j=right-1
        pivot=arr[right]
        while i<j:
            while i<right and arr[i]<pivot:
                i+=1
            while j>=i and arr[j]>=pivot:
                j-=1
            if i<j:
                arr[i],arr[j]=arr[j],arr[i]
        if arr[i]>pivot:
            arr[i],arr[right]=arr[right],arr[i]
        return i
if __name__=='__main__':
    obj=QuickSort()
    testCases = [[],[2, 2, 2, 2], [0,0],[1,1,3,2,-1,-1],[201,492,6,1,0,1,0,3,6],[-1,-7,-1,0,-3,-6],[9,4,-1,5,2,-7]]
    for c,i in enumerate(testCases):
        print("Test Case -",c," ",i, "-->", obj.quickSort(i,0,len(i)-1)) 
```

```python
class QuickSort:
    def quickSort(self,arr, left, right):
        if left < right:
            pos = self.partition(arr, left, right)
            self.quickSort(arr, left, pos - 1)
            self.quickSort(arr, pos + 1, right)
        return arr

    def partition(self, arr, left, right):
        pivot = arr[right]
        i = left - 1

        for j in range(left, right):
            if arr[j] < pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]

        # place pivot correctly
        arr[i + 1], arr[right] = arr[right], arr[i + 1]
        return i + 1
    
    
if __name__=='__main__':
    obj=QuickSort()
    testCases = [[], [0,0],[1,1,3,2,-1,-1],[201,492,6,1,0,1,0,3,6],[-1,-7,-1,0,-3,-6],[9,4,-1,5,2,-7]]
    for c,i in enumerate(testCases):
        print("Test Case -",c," ",i, "-->", obj.quickSort(i, 0, len(i)-1)) 
```

```python
class HeapSort:
    def buildHeap(self,arr):
        n=len(arr)
        for i in range(n//2-1,-1,-1):
            self.heapify(arr,n,i)
        return self.heapSort(arr,n)

    def heapify(self,arr,n,i):
        largest=i
        left=2*i+1
        right=2*i+2
        if(left<n and arr[largest]<arr[left]):
            largest=left
        if(right<n and arr[largest]<arr[right]):
            largest=right
        if(largest!=i):
            arr[largest],arr[i]=arr[i],arr[largest]
            self.heapify(arr,n,largest)

    def heapSort(self,arr,n):
        for i in range(n-1,-1,-1):
            arr[i],arr[0]=arr[0],arr[i]
            self.heapify(arr,i,0)
        return arr
if __name__=="__main__":
    obj=HeapSort()
    testCases = [[], [0,0],[1,1,3,2,-1,-1],[201,492,6,1,0,1,0,3,6],[-1,-7,-1,0,-3,-6],[9,4,-1,5,2,-7]]
    for c,i in enumerate(testCases):
        print("Test Case -",c," ",i, "-->", obj.buildHeap(i))
```

```python
class TreeNode:
    def __init__(self,val=None,left=None,right=None):
        self.val=val
        self.left=left
        self.right=right
class BinaryTree:
    def constructTree(self,arr,root,i,n):
        if i<n:
            root=TreeNode(arr[i])
            root.left=self.constructTree(arr,root.left,2*i+1,n)
            root.right=self.constructTree(arr,root.right,2*i+2,n)
        return root
    def preOrder(self, root, result):
        if not root:
            return result
        result.append(root.val)
        self.preOrder(root.left, result)
        self.preOrder(root.right, result)
        return result
    
obj=BinaryTree()
arr=[3,6,1,8,0,4,5,2]
root = obj.constructTree(arr,None,0,len(arr))
obj.preOrder(root,[])
```
