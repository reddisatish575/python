```python
class Sorting:
    def bubbleSort(self,arr):
        print("Array Before Bubble Sort : ", arr)
        l=len(arr)
        for i in range(l-1):
            isSwapped=False
            for j in range(l-i-1):
                if arr[j]>arr[j+1]:
                    arr[j],arr[j+1]=arr[j+1],arr[j]
                    isSwapped=True
            if not isSwapped:
                return arr
        return arr
obj=Sorting()
arr=[0,-1,0,0,1,-2,5,3,9,7,5,6,4,-999,999,2]
print("Array After Bubble Sort : ", obj.bubbleSort(arr))
```

```python
class Sorting:
    def insertionSort(self,arr):
        print("Array Before Insertion Sort : ", arr)
        l=len(arr)
        for i in range(1,l):
            gap=i
            val=arr[i]
            while gap>0 and arr[gap-1]>val:
                arr[gap]=arr[gap-1]
                gap-=1
            arr[gap]=val
        return arr
obj=Sorting()
arr=[0,-1,0,0,1,-2,5,3,9,7,5,6,4,-999,999,2]
print("Array After Insertion Sort :: ",obj.insertionSort(arr))
```

```python
class Sorting:
    def selectionSort(self,arr):
        print("Array Before Selection Sort : ", arr)
        l=len(arr)
        for i in range(l-1):
            minIndex=i
            for j in range(i+1,l):
                if arr[j]<arr[minIndex]:
                    minIndex=j
            if minIndex!=i:
                arr[i],arr[minIndex]=arr[minIndex],arr[i]
        return arr
obj=Sorting()
arr=[0,-1,0,0,1,-2,5,3,9,7,5,6,4,-999,999,2]
print("Array After Selection Sort :: ",obj.selectionSort(arr))
```
